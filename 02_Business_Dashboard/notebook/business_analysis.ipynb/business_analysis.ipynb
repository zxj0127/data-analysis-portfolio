```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

plt.rcParams["font.sans-serif"] = ["Microsoft YaHei"]
plt.rcParams["axes.unicode_minus"] = False
```

```python
file_path = r"D:\Data-Analysis-Portfolio\02_Business_Dashboard\data\superstore_sales.xlsx"

df = pd.read_excel(file_path)

df.head()
```

```python
# 复制一份数据，避免直接修改原始数据
df_clean = df.copy()

# 删除重复值
df_clean = df_clean.drop_duplicates()

# Postal Code 缺失值用 0 填充
df_clean["Postal Code"] = df_clean["Postal Code"].fillna(0)

# 检查缺失值
df_clean.isnull().sum()
```

```python
df_clean.info()
```

```python
total_sales = df_clean["Sales"].sum()
total_profit = df_clean["Profit"].sum()
total_orders = df_clean["Order ID"].nunique()
total_customers = df_clean["Customer ID"].nunique()
total_quantity = df_clean["Quantity"].sum()
avg_order_value = total_sales / total_orders
profit_rate = total_profit / total_sales

summary = pd.DataFrame({
    "指标": ["总销售额", "总利润", "总订单数", "总客户数", "总销量", "客单价", "利润率"],
    "数值": [
        round(total_sales, 2),
        round(total_profit, 2),
        total_orders,
        total_customers,
        total_quantity,
        round(avg_order_value, 2),
        f"{profit_rate:.2%}"
    ]
})

summary
```

```python
region_sales = df_clean.groupby("Region").agg({
    "Sales": "sum",
    "Profit": "sum",
    "Order ID": "nunique"
}).reset_index()

region_sales.columns = ["地区", "销售额", "利润", "订单数"]

region_sales["利润率"] = region_sales["利润"] / region_sales["销售额"]

region_sales["销售额"] = region_sales["销售额"].round(2)
region_sales["利润"] = region_sales["利润"].round(2)
region_sales["利润率"] = region_sales["利润率"].apply(lambda x: f"{x:.2%}")

region_sales.sort_values("销售额", ascending=False)
```

```python
category_analysis = df_clean.groupby("Category").agg({
    "Sales": "sum",
    "Profit": "sum",
    "Order ID": "nunique"
}).reset_index()

category_analysis.columns = ["品类", "销售额", "利润", "订单数"]

category_analysis["利润率"] = category_analysis["利润"] / category_analysis["销售额"]

category_analysis["销售额"] = category_analysis["销售额"].round(2)
category_analysis["利润"] = category_analysis["利润"].round(2)
category_analysis["利润率"] = category_analysis["利润率"].apply(lambda x: f"{x:.2%}")

category_analysis.sort_values("销售额", ascending=False)
```

```python
top10_products = df_clean.groupby("Product Name").agg({
    "Sales": "sum",
    "Profit": "sum",
    "Quantity": "sum"
}).reset_index()

top10_products.columns = ["商品名称", "销售额", "利润", "销量"]

top10_products = top10_products.sort_values("销售额", ascending=False).head(10)

top10_products["销售额"] = top10_products["销售额"].round(2)
top10_products["利润"] = top10_products["利润"].round(2)

top10_products
```

```python
df_clean["Order Date"] = pd.to_datetime(df_clean["Order Date"])

df_clean["Month"] = df_clean["Order Date"].dt.to_period("M").astype(str)

df_clean[["Order Date", "Month"]].head()
```

```python
monthly_sales = df_clean.groupby("Month").agg({
    "Sales": "sum",
    "Profit": "sum",
    "Order ID": "nunique"
}).reset_index()

monthly_sales.columns = ["月份", "销售额", "利润", "订单数"]

monthly_sales["销售额"] = monthly_sales["销售额"].round(2)
monthly_sales["利润"] = monthly_sales["利润"].round(2)

monthly_sales.head()
```

```python
output_path = r"D:\Data-Analysis-Portfolio\02_Business_Dashboard\output\business_analysis_result.xlsx"

with pd.ExcelWriter(output_path) as writer:
    summary.to_excel(writer, sheet_name="核心指标", index=False)
    region_sales.to_excel(writer, sheet_name="地区分析", index=False)
    category_analysis.to_excel(writer, sheet_name="品类分析", index=False)
    top10_products.to_excel(writer, sheet_name="TOP10商品", index=False)
    monthly_sales.to_excel(writer, sheet_name="月度趋势", index=False)

print("分析结果已保存：", output_path)
```
